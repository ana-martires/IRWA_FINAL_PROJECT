## 2. Inverted Index Construction

We build an inverted index using the processed tokens from Part 1.
Each term is associated with the product IDs (pid) of the documents
that contain it

```python
from collections import defaultdict

def build_inverted_index(df):
    inverted_index = defaultdict(set)

    for _, row in df.iterrows():
        pid = row["pid"]
        tokens = row["search_tokens"]

        for term in tokens:
            inverted_index[term].add(pid)

    return dict(inverted_index)


inverted_index = build_inverted_index(df_processed)

print("Number of indexed terms:", len(inverted_index))
```

## 3. Conjunctive Query Processing (AND)

We implement conjunctive retrieval, where a document is returned
only if it contains all query terms.

Queries are processed using the same normalization pipeline
developed in Part 1

```python
def search_AND(query, inverted_index):
    # Apply the same preprocessing used for documents
    query_terms = preprocess_text(query)

    if not query_terms:
        return set()

    # Retrieve the documents containing each query term
    posting_lists = [
        inverted_index.get(term, set())
        for term in query_terms
    ]

    # AND operation: intersection of all posting lists
    matching_docs = set.intersection(*posting_lists)

    return matching_docs
```

## 5. TF-IDF Ranking

After retrieving documents using conjunctive (AND) queries, we rank
the matching products using TF-IDF

TF-IDF assigns weights to terms based on their frequency within a
document and how common they are across the collection. Then we use cosine similarity between the query vector and document
vectors to rank the retrieved products

```python
import math

def compute_idf(inverted_index, total_docs):
    idf = {}

    for term, posting_list in inverted_index.items():
        document_frequency = len(posting_list)

        idf[term] = math.log(total_docs / document_frequency)

    return idf


idf = compute_idf(inverted_index, len(df_processed))

print("IDF for 'cotton':", round(idf["cotton"], 3))
print("IDF for 'jean':", round(idf["jean"], 3))
```

```python
from collections import Counter

def compute_document_vectors(df, idf):
    doc_vectors = {}

    for _, row in df.iterrows():
        pid = row["pid"]
        tokens = row["search_tokens"]

        term_counts = Counter(tokens)

        vector = {}

        for term, frequency in term_counts.items():
            tf = frequency
            vector[term] = tf * idf[term]

        # L2 normalization
        norm = math.sqrt(sum(weight**2 for weight in vector.values()))

        if norm > 0:
            vector = {
                term: weight / norm
                for term, weight in vector.items()
            }

        doc_vectors[pid] = vector

    return doc_vectors


doc_vectors = compute_document_vectors(df_processed, idf)

print("Document vectors created:", len(doc_vectors))
```

### Query Vectorization and Cosine Similarity

We apply the same preprocessing and TF-IDF weighting to the queries.
Then, we calculate cosine similarity between each query and the documents
retrieved by the AND search

Since both vectors are L2-normalized, cosine similarity can be computed
using their dot product

```python
def compute_query_vector(query, idf):
    query_terms = preprocess_text(query)
    term_counts = Counter(query_terms)

    query_vector = {}

    for term, frequency in term_counts.items():
        if term in idf:
            tf = frequency
            query_vector[term] = tf * idf[term]

    # L2 normalization
    norm = math.sqrt(sum(weight**2 for weight in query_vector.values()))

    if norm > 0:
        query_vector = {
            term: weight / norm
            for term, weight in query_vector.items()
        }

    return query_vector
```

```python
def cosine_similarity(query_vector, doc_vector):
    score = 0

    for term, query_weight in query_vector.items():
        doc_weight = doc_vector.get(term, 0)
        score += query_weight * doc_weight

    return score
```

```python
def rank_documents(query, inverted_index, doc_vectors, idf):
    # First retrieve matching documents using AND
    matching_docs = search_AND(query, inverted_index)

    # Build the query TF-IDF vector
    query_vector = compute_query_vector(query, idf)

    # Calculate similarity for each matching document
    ranked_results = []

    for pid in matching_docs:
        score = cosine_similarity(query_vector, doc_vectors[pid])
        ranked_results.append((pid, score))

    # Sort by score (descending)
    ranked_results.sort(key=lambda x: (-x[1], x[0]))

    return ranked_results
```

## 1. Evaluation metrics implementation

### a) Precision@K

```python
#proportion of relevant docs in the top k retrieved
def precision_at_k(ranked_results, relevant_pids, k):
    if k <= 0:
        return 0.0

    top_k_results = ranked_results[:k]
    relevant_count = 0

    for pid in top_k_results:
        if pid in relevant_pids:
            relevant_count += 1

    precision = relevant_count / k

    return precision
```

### b) Recall@K

```python
#proportion of relevant docs found in the top k vs. the total number of relevant docs.
def recall_at_k(ranked_results, relevant_pids, k):
    if not relevant_pids:
        return 0.0

    top_k_results = ranked_results[:k]
    relevant_count = 0
    for pid in top_k_results:
        if pid in relevant_pids:
            relevant_count += 1

    recall = relevant_count / len(relevant_pids)

    return recall
```

### c) Average Precision @K

```python
#this rewards the system for putting relevant products near the top
def avg_precision_at_k(ranked_results, relevant_pids, k):
    if k <= 0:
        return 0.0

    top_k_results = ranked_results[:k]
    relevant_count = 0
    precision_sum = 0.0

    for i, pid in enumerate(top_k_results):
        if pid in relevant_pids:
            relevant_count += 1
            precision_sum += relevant_count / (i + 1)

    return precision_sum / len(relevant_pids)
```

### d) F1-Score@K

```python
#harmonic mean of precision and recall
def f1_at_k(ranked_results, relevant_pids, k):
    precision = precision_at_k(ranked_results, relevant_pids, k)
    recall = recall_at_k(ranked_results, relevant_pids, k)

    if precision + recall == 0:
        return 0.0

    f1_score = 2 * (precision * recall) / (precision + recall)

    return f1_score
```

### e) Mean Average Precision (MAP)

```python
def mean_avg_precision_at_k(ranked_results, relevant_pids, k):
    if k <= 0:
        return 0.0

    avg_precisions = []

    for query_id, ranked_pids in ranked_results.items():
        avg_precision = avg_precision_at_k(
            ranked_pids, relevant_pids[query_id], k)
        avg_precisions.append(avg_precision)

    return sum(avg_precisions) / len(avg_precisions)
```

### f) Mean Reciprocal Rank (MRR)

```python
#measure how high the first relevant doc appears in ranked result.
def mean_reciprocal_rank_at_k(ranked_results, relevant_pids, k):
    if k <= 0:
        return 0.0

    reciprocal_ranks = []

    for query_id, ranked_pids in ranked_results.items():
        relevant = relevant_pids[query_id]
        reciprocal_rank = 0.0

        for rank, pid in enumerate(ranked_pids[:k], start=1):
            if pid in relevant:
                reciprocal_rank = 1 / rank
                break

        reciprocal_ranks.append(reciprocal_rank)

    return sum(reciprocal_ranks) / len(reciprocal_ranks)
```

### g) Normalised Discounted Cumulative Gain (NDCG)

```python
def ndcg_at_k(ranked_results, relevant_pids, k):
    if k <= 0:
        return 0.0

    top_k = ranked_results[:k]

    dcg = 0.0
    for rank, pid in enumerate(top_k, start=1):
        relevance = 1 if pid in relevant_pids else 0
        dcg += (2**relevance - 1) / math.log2(rank + 1)

    ideal_relevances = [1] * min(len(relevant_pids), k)

    idcg = sum(
        (2**relevance - 1) / math.log2(rank + 1)
        for rank, relevance in enumerate(ideal_relevances, start=1)
    )
    if idcg == 0:
        return 0.0

    return dcg / idcg
```

Each metric measures a different side of search quality. Precision@K shows what percentage of the top K results are correct, which tells us how clean the first page is. Recall@K measures how many of all relevant items in our pool were successfully found in the top K. F1@K combines precision and recall into one balanced score. Average Precision (AP@K) and Mean Average Precision (MAP@K) check if relevant items appear near the top of the list. Mean Reciprocal Rank (MRR@K) measures how fast a user finds the very first correct result. Finally, NDCG@K measures overall ranking quality by giving more credit to relevant items placed at higher positions.

Our test results show clear patterns across the five queries. MRR@10 is 1.0 for every query because the very first result was always relevant. For three queries ("men winter hooded sweatshirt", "women cotton printed kurta", and "men black polo shirt"), Precision@10 and NDCG@10 reached 1.0, meaning all returned top 10 results were correct. The lower Recall@10 for these queries (between 0.37 and 0.45) is not a ranking mistake; it happens because these queries have more than 10 relevant items in total, so a top 10 list can only hold a fraction of them. The query "women black full sleeve round neck top" got a low Precision@10 of 0.30 because the strict AND rule found only 4 items in total. The query "men blue slim fit jeans" had the worst score, with a Precision@10 of 0.40 and NDCG@10 of 0.496. This happened because items like "Tapered Fit" jeans were ranked high just because the word "slim" appeared in their descriptions. Overall, the system reached a MAP@10 of 0.492.

c. Analyze the current search system and identify its main problems or limitations.
For each issue you find, propose possible ways to resolve it. Consider aspects
such as retrieval accuracy, ranking quality, handling of different field types, query
formulation, and indexing strategies.

Our analysis highlights four main problems with the current search system. First, using a strict AND rule means a product must contain every single word from the query. If a product misses just one word, such as "blue" or "cotton", it is left out completely, which drops the recall score. Second, 39.7% of products in the dataset do not have a description, which creates unfair TF-IDF scores compared to items with full descriptions. Third, mixing the title, details, and description into one text field treats all words equally. As a result, a match inside a long description can wrongly push an item above an exact title match. Fourth, basic stemming cannot connect synonyms like "hoodie" and "sweatshirt" or recognize special attributes like color and gender.

We propose three simple improvements to fix these problems. First, we should replace the strict AND rule with BM25 ranking or a soft-AND rule that allows matching most words instead of all of them. This will help find more useful products and increase recall. Second, we should use field-weighted scoring by giving higher importance to matches in the title (60%) and product details (30%) compared to the description (10%). This will stop description noise from messing up the ranking. Third, we should add query expansion using a simple dictionary of synonyms or use vector search with sentence transformers. This will help the engine understand what the user wants even if they use different words.
